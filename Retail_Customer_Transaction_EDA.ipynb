# 4. Data Quality & Cleaning

```python
# Checking missing values in each column
df.isnull().sum()
```

```python
# Checking duplicate rows
df.duplicated().sum()
```

```python
# Displaying duplicate rows
df[df.duplicated(keep=False)].sort_values(by='Transaction_ID')
```

```python
# Creating a copy of the original dataset
clean_df = df.copy()

# Removing duplicate rows
clean_df = clean_df.drop_duplicates()

# Checking the new dataset shape
clean_df.shape
```

```python
# Checking missing values after removing duplicates
clean_df.isnull().sum().sort_values(ascending=False)
```

```python
# RULE => Total_Amount = Amount * Total_Purchases

# Filling missing Total_Amount values
clean_df['Total_Amount'] = clean_df['Total_Amount'].fillna(clean_df['Amount'] * clean_df['Total_Purchases'])

# Filling missing Amount values
clean_df['Amount'] = clean_df['Amount'].fillna(clean_df['Total_Amount'] / clean_df['Total_Purchases'])

# Filling missing Total_Purchases values
clean_df['Total_Purchases'] = clean_df['Total_Purchases'].fillna((clean_df['Total_Amount'] / clean_df['Amount']).round())

# Checking remaining missing values
clean_df[['Amount', 'Total_Purchases', 'Total_Amount']].isnull().sum()
```

```python
clean_df[clean_df[['Amount', 'Total_Purchases', 'Total_Amount']].isnull().any(axis=1)]
```

```python
#Removed the record with missing Amount & Total_Amount values because these essential sales attributes couldn't be reliably calculated or imputed.
clean_df = clean_df.dropna(subset=['Amount', 'Total_Amount'])

clean_df[['Amount', 'Total_Purchases', 'Total_Amount']].isnull().sum()
```

```python
clean_df['Date'] = pd.to_datetime(clean_df['Date'],errors='coerce')

clean_df['Date'].isnull().sum()
```

```python
clean_df['Year'] = clean_df['Year'].fillna(clean_df['Date'].dt.year)

clean_df['Month'] = clean_df['Month'].fillna(clean_df['Date'].dt.month_name())

# Extracting the correct month names from Date
date_month = clean_df['Date'].dt.month_name()

# Checking incorrect month values
valid_dates = clean_df['Date'].notna()

month_mismatches = (
    valid_dates & (clean_df['Month'] != date_month)
)

print("Incorrect Month Values:", month_mismatches.sum())

# Correcting incorrect month values
clean_df.loc[month_mismatches, 'Month'] = date_month[month_mismatches]

# Checking remaining missing and incorrect values
print("Missing Months:", clean_df['Month'].isnull().sum())

print(
    "Remaining Mismatches:",
    (
        clean_df.loc[valid_dates, 'Month'] !=
        date_month[valid_dates]
    ).sum()
)

clean_df[['Date', 'Year', 'Month']].isnull().sum()
```

### Missing Values Handling Decisions

Missing values in the essential sales columns were handled using the mathematical relationship between Amount, Total_Purchases, and Total_Amount.

One record was removed because both Amount and Total_Amount were missing and could not be reliably calculated.

Missing values in other columns were retained because they could not be accurately inferred without making assumptions. These include customer details, contact information, demographic data, and transaction attributes.

Missing Date and Time values were retained because their original values could not be reliably recovered. Year and Month values were filled using Date whenever possible.

Additionally, incorrect Month values were identified by comparing them with Date and corrected using the actual month extracted from the date.

```python
clean_df.dtypes
```

```python
# Converting ID and contact columns to object
clean_df['Transaction_ID'] = clean_df['Transaction_ID'].astype('object')
clean_df['Customer_ID'] = clean_df['Customer_ID'].astype('object')
clean_df['Phone'] = clean_df['Phone'].astype('object')
clean_df['Zipcode'] = clean_df['Zipcode'].astype('object')

# Converting numerical columns to integer
clean_df['Age'] = clean_df['Age'].astype('Int64')
clean_df['Year'] = clean_df['Year'].astype('Int64')
clean_df['Total_Purchases'] = clean_df['Total_Purchases'].astype('Int64')
clean_df['Ratings'] = clean_df['Ratings'].astype('Int64')

clean_df.dtypes
```

```python
# Calculating quartiles and IQR
Q1 = clean_df['Total_Amount'].quantile(0.25)
Q3 = clean_df['Total_Amount'].quantile(0.75)

IQR = Q3 - Q1

# Calculating outlier boundaries
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound.round(2))
print("Upper Bound:", upper_bound.round(2))

# Counting outliers
outliers = clean_df[(clean_df['Total_Amount'] < lower_bound) | (clean_df['Total_Amount'] > upper_bound)]

print("Number of outliers:", len(outliers))
```

```python
# Separating lower and upper outliers
lower_outliers = clean_df[clean_df['Total_Amount'] < lower_bound]
upper_outliers = clean_df[clean_df['Total_Amount'] > upper_bound]

print("Lower outliers:", len(lower_outliers))
print("Upper outliers:", len(upper_outliers))
```

### Outlier Handling Decision

Using the IQR method, 3,905 upper outliers and 0 lower outliers were detected in the Total_Amount column.

These values were retained because unusually high transaction amounts may represent genuine sales rather than data errors. Removing them could result in information loss and distort the sales analysis.

# 5. Data Visualization

```python
category_sales = clean_df.groupby('Product_Category')['Total_Amount'].sum().round()

category_sales
```

```python
plt.figure(figsize=(8, 4))

plt.bar(category_sales.index, category_sales.values)

plt.title('Total Sales by Product Category')
plt.xlabel('Product Category')
plt.ylabel('Total Sales')

plt.gca().yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, pos: f'{x/1e6:.0f}M')
)

plt.show()
```

```python
monthly_sales = clean_df.dropna(subset=['Date']).groupby(
    pd.Grouper(key='Date', freq='MS')
)['Total_Amount'].sum().reset_index()

monthly_sales
```

```python
import matplotlib.dates as mdates

plt.figure(figsize=(9, 4))

plt.plot(
    monthly_sales['Date'],
    monthly_sales['Total_Amount'],
    marker='o',
    label='Monthly Sales'
)

plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')

plt.gca().yaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, pos: f'{x/1e6:.0f}M')
)

plt.xticks(rotation=45)

plt.grid(True)
plt.legend()

plt.gca().xaxis.set_major_locator(mdates.MonthLocator())
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))


plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.hist(clean_df['Total_Amount'], bins=30)

plt.title('Distribution of Transaction Amounts')
plt.xlabel('Total Amount')
plt.ylabel('Frequency')

plt.show()
```

```python
plt.figure(figsize=(8, 4))

plt.scatter(
    clean_df['Amount'],
    clean_df['Total_Purchases']
)

plt.title('Amount vs Total Purchases')
plt.xlabel('Amount')
plt.ylabel('Total Purchases')

plt.show()
```

# 6. Conclusion & Insights

Based on the Exploratory Data Analysis (EDA), the following insights were identified:

1. **Product Category Performance:** Electronics generated the highest total sales, while Home Decor recorded the lowest total sales among all product categories.

2. **Monthly Sales Trends:** Monthly sales remained relatively stable throughout the analyzed period, with some fluctuations. February 2024 recorded the lowest monthly sales.

3. **Sales Distribution:** Most transactions had relatively low to moderate total amounts, while high-value transactions were less frequent.

4. **Relationship Between Variables:** The scatter plot shows no clear visual linear relationship between the unit price (Amount) and the number of items purchased (Total_Purchases).

Overall, the analysis provides useful insights into sales performance, transaction patterns, and product category contributions.
